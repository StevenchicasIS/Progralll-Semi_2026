**Ejercicio** Encontrar la nota final de cada materia,haciendo uso de funciones

```python
def calcular_nota_corte(corte_notas):
    """
    Calcula la nota final de un corte (c1, c2, c3) con las ponderaciones dadas.
    """
    lab1_peso = 0.3
    lab2_peso = 0.3
    parcial_peso = 0.4

    nota_c = (
        corte_notas['lab1'] * lab1_peso +
        corte_notas['lab2'] * lab2_peso +
        corte_notas['parcial'] * parcial_peso
    )
    return nota_c

def calcular_nota_final_materia(materia_notas):
    """
    Calcula la nota final de una materia promediando las notas de los tres cortes.
    """
    notas_cortes = []
    for corte in ['c1', 'c2', 'c3']:
        if corte in materia_notas:
            nota_c = calcular_nota_corte(materia_notas[corte])
            notas_cortes.append(nota_c)

    if notas_cortes:
        return sum(notas_cortes) / len(notas_cortes)
    else:
        return 0 # En caso de que no haya notas de cortes
```

```python
# Iterar sobre cada materia y calcular su nota final
print(f"Notas finales para {persona['Nombre']}:\n")

for materia in persona['materias']:
    nombre_materia = materia['nombre']
    notas_materia = materia['notas']

    nota_final = calcular_nota_final_materia(notas_materia)
    print(f" - {nombre_materia}: {nota_final:.2f}")
```

### Determinar si la Materia está Aprobada o Reprobada

A continuación, se define una función para determinar el estado de aprobación de una materia, y se aplica a las notas finales calculadas.

```python
def determinar_estado_materia(nota_final, nota_minima_aprobacion=6.0):
    """
    Determina si una materia está aprobada o reprobada.
    """
    if nota_final >= nota_minima_aprobacion:
        return "Aprobada"
    else:
        return "Reprobada"

# Iterar nuevamente sobre las materias para mostrar el estado
print(f"\nEstado de las materias para {persona['Nombre']}:\n")

for materia in persona['materias']:
    nombre_materia = materia['nombre']
    notas_materia = materia['notas']

    nota_final = calcular_nota_final_materia(notas_materia)
    estado = determinar_estado_materia(nota_final)

    print(f" - {nombre_materia}: Nota Final = {nota_final:.2f}, Estado = {estado}")
```

Haciendo uso de las estructuras de control, determinar si un numero introducido por el usuario es par, impar o primo

```python
num_str = input("Ingrese un número entero: ")

try:
    num = int(num_str)
except ValueError:
    print("Entrada inválida. Por favor, ingrese un número entero.")
else:
    if num <= 1:
        print(f"El número {num} no es primo ni par ni impar.")
    else:
        # Check for even or odd
        if num % 2 == 0:
            print(f"El número {num} es par.")
        else:
            print(f"El número {num} es impar.")

        # Check for prime
        is_prime = True
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            print(f"El número {num} es primo.")
        else:
            print(f"El número {num} no es primo.")
```

Calcular sobre el sueldo las deducciones por ley ISSS(3%), AFP(6.25%), ISR(Segun tabla) e imprima el total deducciones y sueldo neto (20%)

```python
def calcular_isr(sueldo_gravado):
    """Calcula ISR según tabla progresiva (El Salvador)."""
    if sueldo_gravado <= 472.00:
        return 0
    elif sueldo_gravado <= 895.24:
        return (sueldo_gravado - 472.00) * 0.10
    elif sueldo_gravado <= 2038.10:
        return (sueldo_gravado - 895.24) * 0.20 + 42.32
    elif sueldo_gravado <= 3819.79:
        return (sueldo_gravado - 2038.10) * 0.30 + 270.89
    else:
        return (sueldo_gravado - 3819.79) * 0.30 + 804.68

# Lambdas
iss = lambda s: s * 0.03
afp = lambda s: s * 0.0625

# Cálculo del neto
def sueldo_neto(sueldo_bruto):
    deduccion_iss = iss(sueldo_bruto)
    deduccion_afp = afp(sueldo_bruto)
    base_isr = sueldo_bruto - deduccion_iss - deduccion_afp
    deduccion_isr = calcular_isr(base_isr)
    return sueldo_bruto - deduccion_iss - deduccion_afp - deduccion_isr

# Programa principal
while True:
    try:
        sueldo = float(input("Ingrese su sueldo bruto mensual: $"))
    except ValueError:
        print("Debe ingresar un número.")
    else:
        print(f"ISSS: ${iss(sueldo):.2f}")
        print(f"AFP: ${afp(sueldo):.2f}")
        print(f"ISR: ${calcular_isr(sueldo - iss(sueldo) - afp(sueldo)):.2f}")
        print(f"Sueldo neto: ${sueldo_neto(sueldo):.2f}")

    continuar = input("¿Desea calcular otro sueldo? (s/n): ").lower()
    if continuar != 's':
        print("¡Hasta luego!")
        break
```

**Ejercicio 2 : Pago de agua potable**


```python
# Tarifa de agua
def calcular_pago(metros):
    if metros <= 18:
        return 6.0
    elif metros <= 28:
        return 6.0 + (metros - 18) * 0.45
    else:
        return 6.0 + (28 - 18) * 0.45 + (metros - 28) * 0.65

# Programa
while True:
    try:
        consumo = float(input("Metros cúbicos consumidos: "))
    except ValueError:
        print("Ingrese un valor numérico.")
    else:
        pago = calcular_pago(consumo)
        print(f"Consumo: {consumo} mts, Total a pagar: ${pago:.2f}")

    continuar = input("¿Otra conversión? (s/n): ").lower()
    if continuar != 's':
        print("¡Hasta luego!")
        break
```

```python
from functools import reduce

class estadistica:
    def media(self, serie):
        suma = reduce(lambda acum, num: acum + float(num), serie, 0)
        return suma / len(serie)

    def varianza(self, serie):
        media = self.media(serie)
        suma = reduce(lambda acum, num: acum + (float(num) - media) ** 2, serie, 0)
        return suma / len(serie)

    def desviacion(self, serie):
        return self.varianza(serie) ** 0.5

    def media_armonica(self, serie):
        suma_inversos = reduce(lambda acum, num: acum + 1 / float(num), serie, 0)
        return len(serie) / suma_inversos

    def frecuencia(self, serie):
        def acumular(dic, elemento):
            num = float(elemento)
            dic[num] = dic.get(num, 0) + 1
            return dic
        return reduce(acumular, serie, {})
```

3. Sistema estadístico, calcule: (20%) a. Media aritmética b. Mediana c. Moda d. Varianza e. Desviación estándar f. Desviación típica g. Rango y h. Frecuencia

```python
from functools import reduce
from collections import Counter

class estadistica:

    def media(self, serie):
        suma = reduce(lambda acum, num: acum + float(num), serie, 0)
        return suma / len(serie)

    def mediana(self, serie):
        numeros = sorted([float(num) for num in serie])
        n = len(numeros)
        if n % 2 == 0:
            return (numeros[n//2 - 1] + numeros[n//2]) / 2
        else:
            return numeros[n//2]

    def moda(self, serie):
        numeros = [float(num) for num in serie]
        contador = Counter(numeros)
        max_frec = max(contador.values())
        modas = [k for k, v in contador.items() if v == max_frec]
        return modas[0] if len(modas) == 1 else modas

    def varianza(self, serie):
        media = self.media(serie)
        suma = reduce(lambda acum, num: acum + (float(num) - media) ** 2, serie, 0)
        return suma / len(serie)

    def desviacion(self, serie):
        return self.varianza(serie) ** 0.5

    def rango(self, serie):
        numeros = [float(num) for num in serie]
        return max(numeros) - min(numeros)

    def frecuencia(self, serie):
        numeros = [float(num) for num in serie]
        contador = Counter(numeros)
        return dict(sorted(contador.items()))


serie = input("Ingrese la serie de números separados por comas: ").split(",")

est = estadistica()

print("\nResultados Estadísticos:")
print(f"Media: {est.media(serie):.4f}")
print(f"Mediana: {est.mediana(serie):.4f}")
print(f"Moda: {est.moda(serie)}")
print(f"Varianza: {est.varianza(serie):.4f}")
print(f"Desviación Estándar: {est.desviacion(serie):.4f}")
print(f"Desviación Típica: {est.desviacion(serie):.4f}")
print(f"Rango: {est.rango(serie):.4f}")
print(f"Frecuencia: {est.frecuencia(serie)}")
```
